```python
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv(r"C:\Users\yaagh aiyappa\OneDrive\ドキュメント\csv_files\MinimumLife_Data.csv",
                index_col='PartNo')
df = df.drop('Unnamed: 0', axis=1)
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop('LifeMinimum', axis=1),
                                                   df['LifeMinimum'],
                                                   test_size=0.15,
                                                   random_state = 1)
```

```python
trf1 = ColumnTransformer(transformers=[
    ('scaled', StandardScaler(), slice(0,3))
])
```

```python
%%html
<h1> Linear Regression Model </h1>
```

```python
trf2 = LinearRegression()
```

```python
pipe1 = Pipeline([
    ('trf1', trf1),
    ('trf2',trf2)
])
```

```python
pipe1.fit(x_train, y_train)
```

```python
y_pred1 = pipe1.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
R2_score1 = r2_score(y_test,y_pred1)
R2_score1
```

```python
%%html
<h1> Random Forest Regressor Model </h1>
```

```python
trf3 = RandomForestRegressor()
```

```python
pipe2 = Pipeline([
    ('trf1',trf1),
    ('trf3',trf3)
])
```

```python
pipe2.fit(x_train,y_train)
```

```python
y_pred2 = pipe2.predict(x_test)
```

```python
R2_score2 = r2_score(y_test,y_pred2)
R2_score2
```
