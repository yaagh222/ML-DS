```python
company = soup.find_all('div', class_='companyCardWrapper')
len(company)
```

```python
name=[]
ratings=[]
reviews=[]
l=[]
types=[]
for i in company:
    name.append(i.find('h2').text.strip())
    ratings.append(i.find('div', class_='rating_text').text.strip())
    reviews.append(i.find('span', class_= 'companyCardWrapper__ActionCount').text.strip())
    l.append(i.find('span', class_='companyCardWrapper__interLinking').text.strip().split('|'))
for i in range(0, len(l)):
    types.append(l[i][0])
```

```python
final = pd.DataFrame()
lf=[]
for j in range(1, 30):
    url = f"https://www.ambitionbox.com/list-of-companies?page={j}"
    header = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.163 Safari/537.36"}
    webpage = requests.get(url, headers = header).text
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div', class_='companyCardWrapper')
    name=[]
    ratings=[]
    reviews=[]
    l=[]
    types=[]
    for i in company:
        if type(i) != None:
            name.append(i.find('h2').text.strip())
            ratings.append(i.find('div', class_='rating_text').text.strip())
            reviews.append(i.find('span', class_= 'companyCardWrapper__ActionCount').text.strip())
            l.append(i.find('span', class_='companyCardWrapper__interLinking').text.strip().split('|'))
        else:
            break
    for i in range(0, len(l)):
        types.append(l[i][0])
    d = {'Name':name, "Rating":ratings, "No. of Reviews": reviews, "Type of Company":types}
    df = pd.DataFrame(d)
    final = pd.concat([final, df], ignore_index=True)
final
    
```
